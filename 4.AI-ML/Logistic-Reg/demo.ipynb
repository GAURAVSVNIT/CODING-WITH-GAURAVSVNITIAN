```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
from sklearn.datasets import make_classification
```

```python
## create a dataset
X,y=make_classification(n_samples=1000, n_features=20, n_classes=2, random_state=42)
print(X,y) 
```

```python
pd.DataFrame(X).head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
logistic= LogisticRegression()
logistic.fit(X_train, y_train)
```

```python
y_pred=logistic.predict(X_test)
print(y_pred)
```

```python
# logistic.predict_proba(X_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
score=accuracy_score(y_test, y_pred)
print("Accuracy:", score)
conf_matrix = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:\n", conf_matrix)
print("Classification Report:\n", classification_report(y_test, y_pred))
```

```python
model=LogisticRegression()
penalty=['l1','l2','elasticnet']
C_values=[100,10,1.0,0.1,0.01]
solver=['newton-cg','lbfgs', 'liblinear', 'sag','saga']
```

```python
params=dict(penalty=penalty,C=C_values,solver=solver)
```

```python
## Grid Search CV
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import StratifiedKFold
cv=StratifiedKFold()
grid=GridSearchCV(estimator=model,param_grid=params,scoring='accuracy',cv=cv,n_jobs=1)
```

```python
grid
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred=grid.predict(X_test)
```

```python
score=accuracy_score(y_test, y_pred)
print("Accuracy:", score)
conf_matrix = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:\n", conf_matrix)
print("Classification Report:\n", classification_report(y_test, y_pred))
```

## Randomized SearchCV

```python
model=LogisticRegression()
randomcv=RandomizedSearchCV(estimator=model,param_distributions=params,scoring='accuracy',cv=cv,n_jobs=1,n_iter=10)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Original signal x(t)
t = np.array([-2, -1, 0, 1, 2])
x = np.array([-1, 1, 2, 2, 0])

# Subplot configuration
fig, axs = plt.subplots(3, 2, figsize=(12, 12))
fig.suptitle('Signal Transformations')

# 1. x(t-1) - Right shift by 1
axs[0, 0].stem(t + 1, x, basefmt=" ")
axs[0, 0].set_title(r'$x(t-1)$')
axs[0, 0].grid(True)

# 2. x(2-t) - Time reversal and shift
axs[0, 1].stem(2 - t, x, basefmt=" ")
axs[0, 1].set_title(r'$x(2-t)$')
axs[0, 1].grid(True)

# 3. x(2t + 1) - Scaling and shift
t_scaled = (t - 1) / 2
axs[1, 0].stem(t_scaled, x, basefmt=" ")
axs[1, 0].set_title(r'$x(2t + 1)$')
axs[1, 0].grid(True)

# 4. x(4 - t/2) - Time reversal, scaling, and shift
t_scaled = 4 - t / 2
axs[1, 1].stem(t_scaled, x, basefmt=" ")
axs[1, 1].set_title(r'$x\left(4 - \frac{t}{2}\right)$')
axs[1, 1].grid(True)

# 5. [x(t) + x(-t)]u(t) - Even extension and unit step
x_even = x + x[::-1]
x_even[t < 0] = 0
axs[2, 0].stem(t, x_even, basefmt=" ")
axs[2, 0].set_title(r'$[x(t) + x(-t)]u(t)$')
axs[2, 0].grid(True)

# 6. x(t) * [δ(t + 3/2) - δ(t - 3/2)] - Sampling
x_delta = np.zeros_like(t)
x_delta[np.isclose(t, -1.5)] = -1
x_delta[np.isclose(t, 1.5)] = 0
axs[2, 1].stem(t, x_delta, basefmt=" ")
axs[2, 1].set_title(r'$x(t) \left[\delta\left(t + \frac{3}{2}\right) - \delta\left(t - \frac{3}{2}\right)\right]$')
axs[2, 1].grid(True)

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Original signal
n = np.arange(-4, 5)
x_n = np.array([-1, -0.5, 0.5, 1, 1, 1, 1, 0.5,0])

# 1. x[n - 4]: Right shift by 4 units
n1 = n + 4
x1 = x_n

# 2. x[3 - n]: Time reversal and shift by 3 units
n2 = 3 - n
x2 = x_n[::-1]

# 3. x[3n]: Compression by 3
n3 = n[::3]
x3 = x_n[::3]

# 4. x[3n + 1]: Compression and shift by 1 unit
n4 = (n - 1)[::3]
x4 = x_n[::3]

# 5. x[n] u[3 - n]: Step function limiting to n <= 3
u = (n <= 3).astype(int)
x5 = x_n * u

# 6. x[n - 2] δ[n - 2]: Only at n = 2
x6 = np.zeros_like(x_n)
x6[np.where(n == 2)] = x_n[np.where(n == 2)]

# 7. (1/2) {x[n] + (-1)^n x[n]}: Alternating polarity
x7 = 0.5 * (x_n + np.power(-1, n, dtype=float) * x_n)

# 8. x[(n - 1)^2]: Parabolic mapping
n8 = (n - 1) ** 2
x8 = x_n

# Plotting
plt.figure(figsize=(12, 18))
signals = [(n1, x1, 'x[n-4]'), (n2, x2, 'x[3-n]'), (n3, x3, 'x[3n]'), 
           (n4, x4, 'x[3n+1]'), (n, x5, 'x[n]u[3-n]'), (n, x6, 'x[n-2]δ[n-2]'), 
           (n, x7, '0.5{x[n] + (-1)^n x[n]}'), (n8, x8, 'x[(n-1)^2]')]

for i, (n_i, x_i, title) in enumerate(signals):
    plt.subplot(4, 2, i + 1)
    plt.stem(n_i, x_i, basefmt=" ")
    plt.title(title)
    plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Original function g(t) for (a)
t_a = np.array([0, 1, 2])
g_a = np.array([0, 4, 4])

# Original function g(t) for (b)
t_b = np.array([-1, 0, 0.5, 1])
g_b = np.array([-1, 3, -3, 1])

# Transformations
# (i) g(-t): Reflect horizontally
t_a_neg = -t_a
t_b_neg = -t_b

# (ii) -g(t): Reflect vertically
g_a_neg = -g_a
g_b_neg = -g_b

# (iii) g(t-1): Shift right by 1 unit
t_a_shift = t_a + 1
t_b_shift = t_b + 1

# (iv) g(2t): Compression by factor of 2
t_a_comp = t_a / 2
t_b_comp = t_b / 2

# Plotting
fig, axs = plt.subplots(4, 2, figsize=(12, 16))
titles = ["g(-t)", "-g(t)", "g(t-1)", "g(2t)"]

# Plotting for (a)
for i, (t_trans, g_trans) in enumerate([(t_a_neg, g_a), (t_a, g_a_neg), (t_a_shift, g_a), (t_a_comp, g_a)]):
    axs[i, 0].plot(t_trans, g_trans, marker='o', linestyle='-')
    axs[i, 0].set_title(f"(a) {titles[i]}")
    axs[i, 0].grid(True)

# Plotting for (b)
for i, (t_trans, g_trans) in enumerate([(t_b_neg, g_b), (t_b, g_b_neg), (t_b_shift, g_b), (t_b_comp, g_b)]):
    axs[i, 1].plot(t_trans, g_trans, marker='o', linestyle='-')
    axs[i, 1].set_title(f"(b) {titles[i]}")
    axs[i, 1].grid(True)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.patches import FancyBboxPatch
import numpy as np

# Create figure
fig, ax = plt.subplots(1, 1, figsize=(14, 10))

# Define node positions
nodes = {
    'SIM': (0.5, 0.7),
    'CIT': (0.2, 0.5),
    'BLD': (0.8, 0.5),
    'ECO': (0.5, 0.5),
    'TRN': (0.2, 0.3),
    'EVT': (0.8, 0.3),
    'GEN': (0.5, 0.9),
    'API': (0.5, 0.1),
    'DB': (0.2, 0.1),
    'CACHE': (0.8, 0.1),
    'DASH': (0.5, -0.1)
}

# Node descriptions
descriptions = {
    'SIM': 'Simulation Core\n(C++)',
    'CIT': 'Citizen Module\n(Objects + AI)',
    'BLD': 'Building Module\n(Res/Com/Ind)',
    'ECO': 'Economy Module\n(Resources)',
    'TRN': 'Transportation\n(Roads/Vehicles)',
    'EVT': 'Event/Disaster\n(Random/User)',
    'GEN': 'Procedural\nGenerator',
    'API': 'API Layer\n(WebSocket/REST)',
    'DB': 'PostgreSQL\n(Historical)',
    'CACHE': 'Redis\n(Live Cache)',
    'DASH': 'React Dashboard\n(Visualization)'
}

# Draw nodes
for node, (x, y) in nodes.items():
    # Create fancy box
    box = FancyBboxPatch((x-0.06, y-0.04), 0.12, 0.08,
                         boxstyle="round,pad=0.01",
                         facecolor='lightblue',
                         edgecolor='navy',
                         linewidth=2)
    ax.add_patch(box)
    
    # Add text
    ax.text(x, y, descriptions[node], ha='center', va='center',
            fontsize=8, fontweight='bold')

# Define connections
connections = [
    ('GEN', 'SIM'),
    ('SIM', 'CIT'),
    ('SIM', 'BLD'),
    ('SIM', 'ECO'),
    ('SIM', 'TRN'),
    ('SIM', 'EVT'),
    ('ECO', 'BLD'),
    ('CIT', 'ECO'),
    ('TRN', 'CIT'),
    ('API', 'SIM'),
    ('API', 'DB'),
    ('API', 'CACHE'),
    ('DASH', 'API')
]

# Draw connections
for start, end in connections:
    x1, y1 = nodes[start]
    x2, y2 = nodes[end]
    ax.annotate('', xy=(x2, y2), xytext=(x1, y1),
                arrowprops=dict(arrowstyle='->', lw=1.5, color='darkred'))

# Set plot properties
ax.set_xlim(-0.1, 1.1)
ax.set_ylim(-0.2, 1.0)
ax.set_aspect('equal')
ax.axis('off')
ax.set_title('Modular City Simulation - OOP Architecture', 
             fontsize=16, fontweight='bold', pad=20)

plt.tight_layout()
plt.show()

print("Architecture diagram created with matplotlib!")
```
